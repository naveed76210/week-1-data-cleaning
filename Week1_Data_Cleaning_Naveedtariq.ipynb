Week 1: Applicant Dataset Cleaning and Exploration

Objective: To clean and explore an applicant dataset using Python Pandas.

Tools: Google Colab, Python, Pandas

Tasks:

- Create and load the applicant dataset.
- Explore the dataset and identify data quality issues.
- Handle missing values.
- Remove duplicate records.
- Standardize text values.
- Clean dates and phone numbers.
- Generate a data quality summary.
- Export the cleaned dataset to CSV.

```python
import pandas as pd

names = [
    "Ali Khan", "Ahmed Raza", "Sara Ahmed", "Usman Malik",
    "Ayesha Noor", "Bilal Ahmed", "Fatima Shah", "Hassan Ali",
    "Zainab Iqbal", "Hamza Tariq"
]

domains = [
    "Web Development", "Data Science", "Cyber Security",
    "App Development", "UI/UX Design"
]

universities = [
    "ABC University", "XYZ University", "City University",
    "National University", "Tech University"
]

statuses = ["Approved", "Pending", "Under Review", "Rejected"]

data = []

for i in range(50):
    data.append({
        "Applicant Name": names[i % 10],
        "Email": f"applicant{i+1}@example.com",
        "Phone": f"0300{1000000+i}",
        "Domain Applied": domains[i % 5],
        "University": universities[i % 5],
        "Application Date": f"2026-01-{(i % 28)+1:02d}",
        "Status": statuses[i % 4]
    })

df = pd.DataFrame(data)

extra_rows = [
    ["  Bilal Ahmed ", "bilal@example.com", "0300-1234567",
     "web dev", " ABC University ", "15/01/2026", " pending "],
    ["Fatima Shah", "fatima@example.com", "0311 1234567",
     "DATA SCIENCE", None, "2026-01-16", "APPROVED"],
    ["Hassan Ali", "hassan@example.com", None,
     "cyber security", "XYZ University", "invalid-date", None],
    ["Zainab Iqbal", "zainab@example.com", "03221234567",
     "webdev", "City University", "2026-01-18", "under review"],
    ["Hamza Tariq", "hamza@example.com", "0333-1234567",
     "app development", "National University", "19/01/2026", "REJECTED"],
    [None, "missingname@example.com", "03441234567",
     "UI/UX Design", None, "2026-01-20", "pending"],
    ["Usman Malik", None, "03001234567",
     "Web Development", "Tech University", "2026-01-21", "Approved"],
    ["Ayesha Noor", "ayesha@example.com", "0311-2223333",
     "web development", "ABC University", "2026-01-22", "Pending"],
    ["Sara Ahmed", "sara@example.com", "0322 3334444",
     "Data Science", "XYZ University", "2026-01-23", "Under Review"],
    ["Ali Khan", "applicant1@example.com", "03001000000",
     "Web Development", "ABC University", "2026-01-01", "Approved"]
]

extra_df = pd.DataFrame(extra_rows, columns=df.columns)
df = pd.concat([df, extra_df], ignore_index=True)

df.to_csv("applicants.csv", index=False)

print("Dataset created successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nFirst 5 Records:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nExact Duplicate Rows:")
print(df.duplicated().sum())
```

Day 1: Initial Dataset Exploration

The dataset was successfully created and loaded using Pandas. It contains 60 rows and 7 columns. Initial exploration was performed using "head()", "info()", "isnull().sum()", and "duplicated().sum()".

The dataset will be checked for missing values, duplicate records, inconsistent text formatting, and date-format issues before cleaning.

```python
print("Missing values before cleaning:")
print(df.isnull().sum())
```

```python
df = df.replace(r'^\s*$', pd.NA, regex=True)

print(df.isnull().sum())
```

```python
# Remove rows where Applicant Name is missing
df = df.dropna(subset=["Applicant Name"])

# Fill missing values with sensible defaults
df["Email"] = df["Email"].fillna("not_provided@example.com")
df["University"] = df["University"].fillna("Not Specified")
df["Status"] = df["Status"].fillna("Under Review")

print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nDataset shape after cleaning:", df.shape)
```

Day 2: Handling Missing Values

Missing values were identified using Pandas. Blank or whitespace-only cells were converted to missing values.

Rows with missing Applicant Name were removed because the applicant could not be reliably identified. Missing Email values were filled with a placeholder, while missing University values were replaced with "Not Specified". Missing Status values were set to "Under Review" because the application status was not recorded.

The missing-value counts were checked again after cleaning to verify the results.

```python
print(df.isnull().sum())
```

```python
print("Exact duplicate rows:", df.duplicated().sum())

print("Duplicate emails:", df.duplicated(subset=["Email"]).sum())
```

```python
# Check exact duplicate rows
print("Exact duplicates before cleaning:", df.duplicated().sum())

# Remove exact duplicate rows
df = df.drop_duplicates()

print("Exact duplicates after cleaning:", df.duplicated().sum())
print("Dataset shape:", df.shape)
```

```python
print("Duplicate emails:", df.duplicated(subset=["Email"]).sum())

# Show rows with duplicate emails
display(df[df.duplicated(subset=["Email"], keep=False)])
```

```python
df = df.drop_duplicates(subset=["Email"], keep="first")

print("Duplicate emails after cleaning:",
      df.duplicated(subset=["Email"]).sum())

print("Final dataset shape:", df.shape)
```

```python
print("Exact duplicate rows:", df.duplicated().sum())
print("Duplicate emails:", df.duplicated(subset=["Email"]).sum())
print("Dataset shape:", df.shape)
```

### Day 3: Removing Duplicate Records

Exact duplicate rows were identified and removed using Pandas. Duplicate email addresses were also checked, and repeated applicant records were removed while keeping the first occurrence.

After cleaning, the dataset contained 58 rows and 7 columns. A final verification confirmed that no exact duplicate rows or duplicate email addresses remained.

```python
print("Domain values:")
print(df["Domain Applied"].unique())

print("\nStatus values:")
print(df["Status"].unique())

print("\nUniversity values:")
print(df["University"].unique())
```

```python
text_columns = [
    "Applicant Name",
    "Email",
    "Domain Applied",
    "University",
    "Status"
]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

print("Extra spaces removed.")
```

```python
domain_mapping = {
    "web dev": "Web Development",
    "webdev": "Web Development",
    "web development": "Web Development",
    "data science": "Data Science",
    "cyber security": "Cyber Security",
    "app development": "App Development",
    "ui/ux design": "UI/UX Design"
}

df["Domain Applied"] = (
    df["Domain Applied"]
    .str.lower()
    .map(domain_mapping)
    .fillna(df["Domain Applied"])
)
```

```python
df["Status"] = df["Status"].str.title()

df["Applicant Name"] = df["Applicant Name"].str.title()

df["Email"] = df["Email"].str.lower()

df["University"] = df["University"].str.title()
```

```python
print("Standardized domains:")
print(df["Domain Applied"].unique())

print("\nStandardized statuses:")
print(df["Status"].unique())

print("\nStandardized universities:")
print(df["University"].unique())
```

Day 4: Text Standardization

Text columns were cleaned by removing leading and trailing whitespace. Domain names were standardized using a mapping dictionary to make inconsistent labels consistent. Status, applicant names, and university names were converted to title case, while email addresses were converted to lowercase.

The unique values were checked after cleaning to verify the standardization results.

```python
print("Application Date data type:")
print(df["Application Date"].dtype)

print("\nSample dates:")
print(df["Application Date"].tail(10))

print("\nMissing phone numbers:")
print(df["Phone"].isnull().sum())
```

```python
df["Application Date"] = pd.to_datetime(
    df["Application Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Date data type:", df["Application Date"].dtype)
print("Invalid dates:", df["Application Date"].isnull().sum())
```

```python
df["Phone"] = df["Phone"].str.replace(
    r"[\s-]", "", regex=True
)

print(df["Phone"].tail(10))
print("Missing phone numbers:", df["Phone"].isnull().sum())
```

```python
print(df[["Application Date", "Phone"]].tail(10))
print("\nData types:")
print(df[["Application Date", "Phone"]].dtypes)
```

```python
print("Invalid/missing dates:", df["Application Date"].isnull().sum())
print("Missing phone numbers:", df["Phone"].isnull().sum())

display(
    df[df["Application Date"].isnull()][
        ["Applicant Name", "Application Date"]
    ]
)
```

Day 5: Data Types, Dates, and Phone Numbers

Application dates were converted to Pandas datetime format. Invalid date values were converted to missing dates using "errors="coerce"" so they could be identified for further review.

Phone numbers were standardized by removing spaces and hyphens. Missing phone numbers were preserved rather than replaced with invented values. The updated data types and cleaned values were checked after processing.

```python
print("DATA QUALITY SUMMARY")
print("-" * 30)

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

print("\nMissing values per column:")
print(df.isnull().sum())

print("\nExact duplicate rows:", df.duplicated().sum())

print(
    "Duplicate emails:",
    df.duplicated(subset=["Email"]).sum()
)

print("\nUnique domains:")
print(df["Domain Applied"].nunique())

print("\nUnique statuses:")
print(df["Status"].nunique())

print("\nColumn data types:")
print(df.dtypes)
```

```python
print("Rows with missing dates:")
display(df[df["Application Date"].isnull()])

print("Rows with missing phone numbers:")
display(df[df["Phone"].isnull()])
```

```python
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

print("\nMissing values per column:")
print(df.isnull().sum())

print("\nExact duplicate rows:", df.duplicated().sum())
print("Duplicate emails:", df.duplicated(subset=["Email"]).sum())
```

Day 6: Data Quality Summary

The cleaned dataset contains 58 rows and 7 columns. No missing values remain in Applicant Name, Email, Domain Applied, University, or Status.

One phone number is missing, and one application date is invalid. No exact duplicate rows or duplicate email addresses remain. The dataset was reviewed to verify the results of the cleaning process before export.
